```python
%sql

SELECT
    COUNT(*) AS total_bronze_events,
    COUNT(DISTINCT CONCAT(topic, '-', partition, '-', offset)) AS distinct_kafka_events
FROM dev.bronze.raw_events;
```

```python
%sql

SELECT
    topic,
    COUNT(*) AS event_count,
    MIN(offset) AS min_offset,
    MAX(offset) AS max_offset
FROM dev.bronze.raw_events
GROUP BY topic
ORDER BY topic;
```
